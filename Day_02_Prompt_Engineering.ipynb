```python
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI

GOOGLE_API_KEY = userdata.get("GEMINI_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=GOOGLE_API_KEY,
    temperature=0
)

print("LLM initialized successfully")
```

Vague Prompt

```python
text = """
FastAPI is a modern Python web framework for building APIs.
It uses Python type hints and provides automatic API documentation.
It supports asynchronous programming and is commonly used for
backend services and microservices.
"""

response = llm.invoke(
    f"Tell me about this text:\n{text}"
)

print(response.content)
```

Make the prompt specific

```python
text = """
FastAPI is a modern Python web framework for building APIs.
It uses Python type hints and provides automatic API documentation.
It supports asynchronous programming and is commonly used for
backend services and microservices.
"""

response = llm.invoke(
    f""" Summarize the following text in 3 bullet points.

TEXT:
{text}
"""
)

print(response.content)
```

Add constraints

```python
response = llm.invoke(
    f"""
Summarize the following text.

Rules:
- Return exactly 3 bullet points.
- Each bullet must contain one sentence.
- Do not add information that is not present in the text.
- Focus only on technical information.

TEXT:
{text}
"""
)

print(response.content)
```

Add a role to the prompt

```python
response = llm.invoke(
    f"""
You are a senior backend engineer reviewing technical documentation.

TASK:
Summarize the important technical information.

RULES:
- Return exactly 3 bullet points.
- Each bullet must contain one sentence.
- Do not add information that is not present in the text.
- Focus only on technical information.

TEXT:
{text}
"""
)

print(response.content)
```

Control the output format

```python
response = llm.invoke(
    f"""
You are a senior backend engineer.

Analyze the following technical text.

Return your answer using exactly this format:

Summary:
<one sentence>

Key Points:
- <point 1>
- <point 2>
- <point 3>

Technical Topics:
<comma-separated list>

TEXT:
{text}
"""
)

print(response.content)
```

Give the model a negative constraint

```python
text = """
FastAPI is a Python web framework used for building APIs.
It uses Python type hints and provides automatic API documentation.
"""

response = llm.invoke(
    f"""
You are a technical documentation assistant.

Answer the question using ONLY the information in the provided text.

Rules:
- Do not use outside knowledge.
- Do not make assumptions.
- If the answer is not contained in the text, say:
  "The provided text does not contain enough information."

TEXT:
{text}

QUESTION:
Who created FastAPI?
"""
)

print(response.content)
```

Separate instructions from data

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
        """
You are a technical documentation assistant.

Follow these rules:
- Answer using only the provided context.
- Do not use outside knowledge.
- Do not invent information.
- If the context does not contain the answer, say:
  "The provided context does not contain enough information."
"""
        ),
        (
            "human",
        """
CONTEXT:
{context}

QUESTION:
{question}
"""
        )
    ]
)
```

```python
context = """
FastAPI is a Python web framework used for building APIs.
It uses Python type hints and provides automatic API documentation.
"""

question = "Who created FastAPI?"

messages = prompt.invoke({
    "context": context,
    "question": question
})

response = llm.invoke(messages)

print(response.content)
```
